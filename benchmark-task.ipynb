```python
import kaggle_benchmarks as kbench

@kbench.task(name="chess mate in 2 puzzle")
def chess_mate_in_2_puzzle(llm):
    fen = "3q2r1/4n2k/p1p1rBpp/PpPpPp2/1P3P1Q/2P3R1/7P/1R5K w - - 1 0"
    board_ascii = """
. . . q . . r .
. . . . n . . k
p . p . r B p p
P p P p P p . .
. . P . . P . Q
. . . P . . R .
. . . . . . . P
. R . . . . . K
"""
    prompt = f"""
You are a chess AI with a chess skill above 1800 ELO rating.

Position (FEN):
{fen}

Board (ASCII):
{board_ascii}

Which move leads to mate in 2?

A) Qxh6+
B) Bg7

Respond ONLY with the option letter (e.g., A) or the exact move notation (e.g., Qxh6+). Do not include any other text.
"""
    response = llm.prompt(prompt).strip()

    correct_answers = ["A", "Qxh6+", "A) Qxh6+"]

    kbench.assertions.assert_in(
        response,
        correct_answers,
        expectation=f"The LLM should identify the correct mate in 2 move. Expected one of {correct_answers}, got '{response}'."
    )

chess_mate_in_2_puzzle.run(kbench.llm)
```
